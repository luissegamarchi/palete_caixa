```python
import pulp as plp
import matplotlib.pyplot as plt
from itertools import product
```

```python
comprimentos_palete = [8, 50, 50]
larguras_palete = [6, 50, 40]

comprimentos_caixa = [3, 5, 7]
larguras_caixa = [2, 4, 6]
```

```python
indice = 0

palete_comprimento = comprimentos_palete[indice]
palete_largura = larguras_palete[indice]

caixa_comprimento = comprimentos_caixa[indice]
caixa_largura = larguras_caixa[indice]

print(f"Dimensões do Palete: {palete_comprimento} x {palete_largura}")
print(f"Dimensões da Caixa: {caixa_comprimento} x {caixa_largura}")
```

```python
dimensoes_palete = [palete_comprimento, palete_largura]

L, W = sorted(dimensoes_palete, reverse=True)
print(f"Dimensões Classificadas do Palete: {L} x {W}")
```

```python
dimensoes_caixa = [caixa_comprimento, caixa_largura]

l_um, w_um = sorted(dimensoes_caixa, reverse=True)
print(f"Dimensões Classificadas da Caixa Deitada: {l_um} x {w_um}")
```

```python
l_dois, w_dois = dimensoes_caixa[::-1]
print(f"Dimensões Classificadas da Caixa em Pé: {l_dois} x {w_dois}")
```

```python
def plotar_caixa(bottom_left: list, 
                 l: int,
                 w: int,
                 color: str,
                 label: str):
    
    x0, y0 = bottom_left
    plt.scatter(x=x0, y=y0, color=color)
    plt.plot([x0, x0 + l], [y0, y0], color=color)
    plt.plot([x0, x0], [y0, y0 + w], color=color)
    plt.plot([x0, x0 + l], [y0 + w, y0 + w], color=color)
    plt.plot([x0 + l, x0 + l], [y0, y0 + w], color=color)
    plt.text(x=x0 + l/2, y=y0 + w/2, s=label, color=color, fontsize=9, ha="center", va="center")

def plotar_palete(L: int, W:int):
    plt.figure(figsize=(8, 6))
    plt.plot([0, L], [0, 0], "k-")
    plt.plot([0, 0], [0, W], "k-")
    plt.plot([0, L], [W, W], "k-")
    plt.plot([L, L], [0, W], "k-")

plotar_palete(L, W)
plotar_caixa(bottom_left=[2, 1], l=3, w=2, color='blue', label='1')
plotar_caixa(bottom_left=[6, 0], l=2, w=3, color='red', label='2')
```

```python
f_um = lambda n_um, l_um, n_dois, l_dois: \
    n_um * l_um + n_dois * l_dois

f_dois = lambda n_um, w_um, n_dois, w_dois: \
    n_um * w_um + n_dois * w_dois
```

```python
print(f_um(n_um=2, l_um=3, n_dois=1, l_dois=2))  # p <= L - w_um
```

```python
print(f_dois(n_um=2, w_um=2, n_dois=1, w_dois=3))  # q <= W - w_um
```

```python
H = []

for n_dois, n_um in product(range(L), range(L)):
    # print(n_um, n_dois)
    p = f_um(n_um, l_um, n_dois, l_dois)
    # print(n_um, n_dois, p)
    if p <= L - w_um:
        if p not in H:
            H.append(p)

        plotar_palete(L, W)

        for n in range(n_um):
            plotar_caixa(bottom_left=[n * l_um, 0], l=l_um, w=w_um,
                          color='black', label=n+1)

        for n in range(n_dois):
            plotar_caixa(bottom_left=[n_um * l_um + n * l_dois, 0], l=l_dois, w=w_dois, color='blue', label=n+1)


        plt.axvline(x=p, color="red", linestyle="--")
        plt.title(f"{n_um = } | {n_dois = } | {p = }\n{H = }")
        plt.grid()            
```

```python
V = []

for n_dois, n_um in product(range(L), range(L)):
    # print(n_um, n_dois)
    q = f_dois(n_um, w_um, n_dois, w_dois)
    # print(n_um, n_dois, q)
    if q <= W - w_um:
        if q not in V:
            V.append(q)

        plotar_palete(L, W)

        for n in range(n_um):
            plotar_caixa(bottom_left=[0, n * w_um], l=l_um, w=w_um,
                          color='black', label=n+1)

        for n in range(n_dois):
            plotar_caixa(bottom_left=[0, n_um * w_um + n * w_dois], l=l_dois, w=w_dois, color='blue', label=n+1)


        plt.axhline(y=q, color="red", linestyle="--")
        plt.title(f"{n_um = } | {n_dois = } | {q = }\n{V = }")
        plt.grid() 
```

```python
H = sorted(H)
V = sorted(V)

print(H)
print(V)
```

```python
lista_rs = list(product(H, V))
display(lista_rs)
```

```python
if indice == 0:
    plotar_palete(L, W)
    for r, s in lista_rs:
        plt.scatter(x=r, y=s, color="purple")

    plt.title("Possibilidades de bottom-left para as caixas")
    plt.grid()
    plt.show()
```

```python
ls = [l_um, l_dois]
ws = [w_um, w_dois]
```

```python
i = 1
print(ls[i-1], ws[i-1])
```

```python
plotar_palete(L, W)

lista_variaveis = []

for (p, q), i in product(lista_rs, [1, 2]):
    if p + ls[i-1] <= L and q + ws[i-1] <= W: # checa se a caixa cabe no palete
        lista_variaveis.append((i, p, q))

        if indice == 0:
            cor = "blue" if i == 1 else "red"
            plt.scatter(x=p, y=q, color=cor) # plota o bottom-left da caixa no palete

            if i == 1:
                plt.text(x=p - 0.3, y=q + 0.15, s=f"({i}, {p}, {q})", color=cor, fontsize=10)
            else:
                plt.text(x=p - 0.3, y=q - 0.25, s=f"({i}, {p}, {q})", color=cor, fontsize=10)
```

```python
print(len(lista_variaveis))
```

```python
prob = plp.LpProblem(name="Palete-Caixa",
                     sense=plp.LpMaximize)

prob
```

```python
dict_variaveis = plp.LpVariable.dicts(name="C", indices=lista_variaveis, lowBound=0, upBound=1, cat=plp.LpBinary)
display(dict_variaveis)
```

```python
# dict_variaveis.values()
sum(dict_variaveis.values())
```

```python
prob += sum(dict_variaveis.values())

display(prob)
```

```python
print(prob.numVariables())
print(prob.numConstraints())
```

```python
def ponto_dentro_caixa(ponto_rs: tuple, ponto_pq: tuple, l: int, w: int) -> bool:
    r, s = ponto_rs
    p, q = ponto_pq
    return p <= r < p + l and q <= s < q + w


print(ponto_dentro_caixa((6,2), (6, 0), l=ls[0], w=ws[0]))
```

```python
plotar_palete(L, W)
r, s = lista_rs[9]
plt.scatter(x=r, y=s, color='blue', s=100)

caixas_sobrepostas = []
for i, p, q in lista_variaveis:
    if  ponto_dentro_caixa((r, s), (p, q), l=ls[i-1], w=ws[i-1]):
        plotar_caixa((p, q), ls[i-1], ws[i-1], color='blue' if  i == 1 else "red", label=f"{i, p, q}")
```

```python

for r, s in lista_rs:
    caixas_sobrepostas = []
    for i, p, q in lista_variaveis:
        if  ponto_dentro_caixa((r, s), (p, q), l=ls[i-1], w=ws[i-1]):
            caixas_sobrepostas.append(dict_variaveis[(i, p, q)])
    prob += sum(caixas_sobrepostas) <= 1, f"Restricao 1.{n}"
    n += 1
```

```python
display(prob)
```

```python
print(prob.numVariables())
print(prob.numConstraints())
```

```python
qtd_maxima_caixas = (L * W) // (l_um * w_um)
print(qtd_maxima_caixas)
```

```python
prob += sum(dict_variaveis.values()) <= qtd_maxima_caixas, "Restricao 2.0"
display(prob)
```

```python
solver = plp.getSolver("PULP_CBC_CMD", timeLimit=None, msg=True)
```

```python
prob.solve(solver)
```
